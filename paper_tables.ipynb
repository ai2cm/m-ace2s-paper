# Masked-training paper tables

Each table is written to `tables/<name>.tex` (LaTeX), `tables/<name>.md` and `tables/<name>.csv` (unrounded values); `TABLES.md` collects them.

```python
from IPython.display import Markdown, display

import tables as T
from plotting import load_data

data = load_data()
TABLES = []


def show(table):
    T.save(table)
    TABLES.append(table)
    display(Markdown(table.md))
```

## `table:t7_trend_1deg`, `table:t7_trend_4deg` — $T_7$ bias and trend error, training vs test period

Annual global-mean $T_7$ of the 46-year rollouts (the series of the annual-mean $T_7$ figures) for the ACE2S and M-ACE2S models, with and without GMR; one table per resolution. Columns: RMSE, mean bias and linear-trend error over 1979–2008 and 2015–2024.

```python
for res in ["1°", "4°"]:
    show(T.trend_table(data, "air_temperature_7", res))
```

## `table:twp_trend_1deg`, `table:twp_trend_4deg` — $TWP$

The same tables for total water path $TWP$.

```python
for res in ["1°", "4°"]:
    show(T.trend_table(data, "total_water_path", res))
```

## `table:configurations` — trained configurations

One row per experiment group, parsed from the training configs in `configs/`. `tables.config_rows` asserts that every config falls in one group and that none has an anthropogenic-forcing input channel.

```python
show(T.config_table())
```
